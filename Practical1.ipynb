```python
a = 10
b = 5.5
name = "Python"
is_valid = True
print("Variables and Data Types:")
print("a =", a, "Type:", type(a))
print("b =", b, "Type:", type(b))
print("name =", name, "Type:", type(name))
print("is_valid =", is_valid, "Type:", type(is_valid))

# Arithmetic Operators
print("\nArithmetic Operators:")
print("Addition:", a + 5)
print("Subtraction:", a - 5)
print("Multiplication:", a * 5)
print("Division:", a / 5)
print("Floor Division:", a // 3)
print("Modulus:", a % 3)
print("Exponent:", a ** 2)

# Assignment Operators
print("\nAssignment Operators:")
x = 10
x += 5
print("x += 5:", x)
x -= 2
print("x -= 2:", x)
x *= 2
print("x *= 2:", x)
x /= 2
print("x /= 2:", x)

# Comparison Operators
print("\nComparison Operators:")
print("a == 10:", a == 10)
print("a != 5:", a != 5)
print("a > 5:", a > 5)
print("a < 20:", a < 20)
print("a >= 10:", a >= 10)
print("a <= 10:", a <= 10)

# Logical Operators
print("\nLogical Operators:")
print("True and False:", True and False)
print("True or False:", True or False)
print("not True:", not True)

# Membership Operators
print("\nMembership Operators:")
text = "Python"
print("'P' in text:", 'P' in text)
print("'z' not in text:", 'z' not in text)

# Identity Operators
print("\nIdentity Operators:")
p = [1, 2, 3]
q = p
r = [1, 2, 3]
print("p is q:", p is q)
print("p is r:", p is r)
print("p is not r:", p is not r)

# Bitwise Operators
print("\nBitwise Operators:")
m = 10
n = 4
print("m & n:", m & n)
print("m | n:", m | n)
print("m << 1:", m << 1)
print("m >> 1:", m >> 1)
```
