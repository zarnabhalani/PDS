```python
import pandas as pd
marks = pd.Series([85, 72, 90, 65, 78])
names = ["A", "B", "C", "D", "E"]
print("Number series:",marks)
data = {
    "Name": ["Zarna", "Bansi", "Khushi", "Krinal"],
    "Department": ["IT", "CE", "CE", "IT"],
    "Marks": [85, 72, 90, 65],
    "Age": [20, 21, 20, 22]
}
df = pd.DataFrame(data)
print("\nDataFrame:")
print(df)
print("\nData Selection:")
print(df["Name"])
print(df.iloc[0:3])
print(df[["Name", "Marks"]])
print("\nData Filtering:")
print(df[df["Marks"] > 70])
print(df[df["Department"] == "CE"])
print("\nGrouping:")
print(df.groupby("Department")["Marks"].mean())
print("\nAggregation:")
print(df.groupby("Department")["Marks"].sum())
```
