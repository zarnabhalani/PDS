```python
import csv
file=open("student.csv","w", newline="")
writer=csv.writer(file)
writer.writerow(["Roll No", "Name", "Marks"])
writer.writerow([1, "Zarna", 85])
writer.writerow([2, "Bansi", 78])
writer.writerow([3, "Krinal", 92])
file.close()
```

```python
file=open("student.csv","r")
reader=csv.reader(file)
print("CSV File Content:")
for row in reader:
    print(row)
file.close()
```

```python
file=open("student.csv", "a", newline="")
writer=csv.writer(file)
writer.writerow([4, "Jensi", 80])
file.close()
print("Data appended")
```

```python
file=open("student.csv", "r")
print("After appending:")
reader=csv.reader(file)
for row in reader:
    print(row)
file.close()
```
