```python
# List
marks = [75, 82, 68, 90]
print("Original list:", marks)
marks.append(85)
print("After append:", marks)
marks.remove(68)
print("After remove:", marks)
marks.sort()
print("After sorting:", marks)

# Tuple
subjects = ("Python", "DBMS", "CN", "WAD")
print("Tuple:", subjects)
print("Number of subjects:", len(subjects))
print("Index of DBMS:", subjects.index("DBMS"))

# Set
numbers = {10, 20, 30, 40}
print("Original set:", numbers)
numbers.add(50)
print("After adding 50:", numbers)
numbers.remove(20)
print("After removing 20:", numbers)
other = {30, 40, 60}
print("Union:", numbers.union(other))
print("Intersection:", numbers.intersection(other))

# Dictionary
student = {
    "name": "Zarna",
    "age": 20,
    "branch": "Computer Engineering"
}
print("Student details:", student)
student["age"] = 21
print("After updating age:", student)
student["city"] = "Vadodara"
print("After adding city:", student)
print("Keys:", student.keys())
print("Values:", student.values())

# String
name = "Python Programming"
print("Original string:", name)
print("Uppercase:", name.upper())
print("Lowercase:", name.lower())
print("Length:", len(name))
print("Replace:", name.replace("Python", "Java"))
print("Count of 'm':", name.count("m"))
```
